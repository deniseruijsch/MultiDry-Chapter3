```python
from pathlib import Path

base_dir = Path("/projects/prjs2023/chapter3/lpjml_output/2026_04_08_run_for_10_locations")

for pft_dir in base_dir.iterdir():
    if not pft_dir.is_dir():
        continue

    print(f"Processing PFT: {pft_dir.name}")

    for i in range(1, 11):
        patch_dir = pft_dir / f"patch_{i}"
        output_dir = patch_dir / "output"

        patch_dir.mkdir(parents=True, exist_ok=True)
        output_dir.mkdir(parents=True, exist_ok=True)

        print(f"  ensured {output_dir}")
```
