```python
import os
import json
```

```python

pft_names_main = [
    # "tropical broadleaved evergreen tree",
    # "tropical broadleaved raingreen tree",
    # "temperate needleleaved evergreen tree",
    # "temperate broadleaved evergreen tree",
    "temperate broadleaved summergreen tree",
    "boreal needleleaved evergreen tree",
    "boreal broadleaved summergreen tree",
    "boreal needleleaved summergreen tree"
]
# Base directories
hotspots_base = "/projects/prjs2023/chapter3/lpjml_input/run_for_10_locations"  # where we saved the 10 patches
lpj_input_base = "/home/druijsch/LPJmL5/dev15oct2025/config_files/2026_04_08_run_for_10_locations"

# PFT list
pfts = pft_names_main  # same as before

# Global LPJmL files
global_files = {
    "no3deposition": "/projects/prjs2023/chapter3/lpjml_input/ndep-noy_histsoc_monthly_1901_2021_units.nc",
    "nh4deposition": "/projects/prjs2023/chapter3/lpjml_input/ndep-nhx_histsoc_monthly_1901_2021_units.nc",
    "soilpH": "/projects/prjs2023/chapter3/lpjml_input/hwsd_soil_data_all_land_oceanfilled.nc",
    "co2": "/projects/prjs2023/chapter3/lpjml_input/co2_1832-2023.dat",
    "landuse": "/projects/prjs2023/chapter3/lpjml_input/MAPSPAM_2020_crops_filled_ESA_01deg_v7.nc",
    "countrycode": "/projects/prjs2023/chapter3/lpjml_input/regridded_cow_full_2018_5min_reduced_filled.nc",
    "landcover": "/projects/prjs2023/chapter3/lpjml_input/2025_04_87_fpc_from_own_landcover_ESA_2019_overwrite_working.nc"
}

# Variable names for LPJmL
var_map = {
    "soil": "soilcode",
    "wind": "sfcwind",
    "temp": "tas",
    "prec": "pr",
    "lwdown": "rlds",
    "swdown": "rsds",
    "vpd": "VPD"
}

# ----------------------------
# LOOP OVER PFTs & PATCHES
# ----------------------------
for pft_name in pfts:
    pft_folder = os.path.join(hotspots_base, pft_name.replace(" ", "_"))
    patches = sorted([d for d in os.listdir(pft_folder) if d.startswith("patch_")])

    for patch in patches:
        patch_path = os.path.join(pft_folder, patch)
        out_dir = os.path.join(lpj_input_base, pft_name.replace(" ", "_"))
        os.makedirs(out_dir, exist_ok=True)

        # LPJmL input dict
        cjson_dict = {}

        # Hotspot-specific files
        for var, var_name in var_map.items():
            file_path = os.path.join(patch_path, f"{var}.nc")
            if os.path.exists(file_path):
                cjson_dict[var] = {"fmt": "cdf", "var": var_name, "name": file_path}

        # Global files
        cjson_dict["no3deposition"] = {"fmt": "cdf", "var": "noy", "name": global_files["no3deposition"]}
        cjson_dict["nh4deposition"] = {"fmt": "cdf", "var": "nhx", "name": global_files["nh4deposition"]}
        cjson_dict["soilpH"] = {"fmt": "cdf", "var": "soil_ph", "name": global_files["soilpH"]}
        cjson_dict["co2"] = {"fmt": "txt", "name": global_files["co2"]}
        cjson_dict["landuse"] = {"fmt": "cdf", "var": "CFTfrac", "name": global_files["landuse"]}
        cjson_dict["countrycode"] = {"fmt": "cdf", "var": "country", "name": global_files["countrycode"]}
        cjson_dict["landcover"] = {"fmt": "cdf", "var": "FPC", "name": global_files["landcover"]}

        # Save as JSON
        cjson_file = os.path.join(out_dir, f"{pft_name.replace(' ','_')}_patch{patch.split('_')[1]}_input.cjson")
        with open(cjson_file, "w") as f:
            json.dump(cjson_dict, f, indent=2)

        print(f"✅ Created {cjson_file}")
```
