```python

pft_names_main = [
    "tropical broadleaved evergreen tree",
    "tropical broadleaved raingreen tree",
    "temperate needleleaved evergreen tree",
    "temperate broadleaved evergreen tree",
    "temperate broadleaved summergreen tree",
    "boreal needleleaved evergreen tree",
    "boreal broadleaved summergreen tree",
    "boreal needleleaved summergreen tree"
]
```

```python
import os

# ----------------------------
# PATHS
# ----------------------------
config_template = "/home/druijsch/LPJmL5/dev15oct2025/config_files/2026_04_08_run_for_10_locations/temperate_needleleaved_evergreen_tree/temperate_needleleaved_evergreen_tree_config_spinup.cjson"

input_base   = "/home/druijsch/LPJmL5/dev15oct2025/config_files/2026_04_08_run_for_10_locations"
output_base  = "/projects/prjs2023/chapter3/lpjml_output/2026_04_08_run_for_10_locations"
restart_base = "/home/druijsch/LPJmL5/dev15oct2025/restart/2026_04_08_run_for_10_locations"

# ----------------------------
# LOAD TEMPLATE ONCE
# ----------------------------
with open(config_template, "r") as f:
    template_text = f.read()

# ----------------------------
# LOOP OVER PFTs
# ----------------------------
for pft_name in pft_names_main:

    pft_clean = pft_name.replace(" ", "_")

    # NEW structure
    pft_base_dir   = os.path.join(input_base, pft_clean, "spinup", "configs")
    pft_input_dir  = pft_base_dir
    pft_config_dir = pft_base_dir

    # create directory tree
    os.makedirs(pft_base_dir, exist_ok=True)

    # ----------------------------
    # FIND INPUT FILES
    # ----------------------------
    input_files = sorted([
        f for f in os.listdir(pft_input_dir)
        if f.endswith("_input.cjson")
    ])

    if not input_files:
        print(f"⚠️ No input files found in {pft_input_dir}")
        continue

    # ----------------------------
    # LOOP OVER PATCHES
    # ----------------------------
    for input_file in input_files:

        try:
            patch_id = input_file.split("_patch")[1].split("_")[0]
        except IndexError:
            print(f"⚠️ Could not parse patch ID from {input_file}")
            continue

        # ----------------------------
        # DEFINE PATHS
        # ----------------------------
        input_path = os.path.join(pft_input_dir, input_file)

        outpath = os.path.join(output_base, pft_clean, f"patch_{patch_id}")
        restart_dir = os.path.join(restart_base, pft_clean, f"patch_{patch_id}")

        os.makedirs(outpath, exist_ok=True)
        os.makedirs(restart_dir, exist_ok=True)

        restart_in   = os.path.join(restart_dir, "restart_1901_nv_stdfire.lpj")
        restart_out  = os.path.join(restart_dir, "restart_1901_nv_stdfire.lpj")
        restart_out2 = os.path.join(restart_dir, "restart_1960_crop_stdfire.lpj")

        # ----------------------------
        # MODIFY TEMPLATE
        # ----------------------------
        new_text = template_text

        # replace INPUT include
        new_text = new_text.replace(
            '#include "/home/druijsch/LPJmL5/dev15oct2025/config_files/2026_03_30_region_droughtmort/2026_03_30_region_droughtmort_input.cjson"',
            f'#include "{input_path}"'
        )

        # replace OUTPATH
        new_text = new_text.replace(
            '"outpath" : "/projects/prjs2023/chapter3/lpjml_output/2026_03_30_region_droughtmort"',
            f'"outpath" : "{outpath}"'
        )

        # replace restart input
        new_text = new_text.replace(
            '/home/druijsch/LPJmL5/dev15oct2025/restart/2026_01_29_baseline_run/restart_1901_nv_stdfire.lpj',
            restart_in
        )

        # replace restart output 1
        new_text = new_text.replace(
            '/home/druijsch/LPJmL5/dev15oct2025/restart/2026_03_30_region_droughtmort/restart_1901_crop_nofire_monthlybackground_droughtmort.lpj',
            restart_out
        )

        # replace restart output 2
        new_text = new_text.replace(
            "/home/druijsch/LPJmL5/dev15oct2025/restart/2026_03_30_region_droughtmort/restart_1901_crop_nofire_monthlybackground_droughtmort2.lpj",
            restart_out2
        )

        # ----------------------------
        # SAVE CONFIG
        # ----------------------------
        out_config = os.path.join(
            pft_config_dir,
            f"{pft_clean}_patch{patch_id}_config_restart.cjson"
        )

        with open(out_config, "w") as f:
            f.write(new_text)

        print(f"✅ Created config: {out_config}")
```
